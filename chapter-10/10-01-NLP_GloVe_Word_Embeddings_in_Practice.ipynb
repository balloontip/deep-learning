Learn how to use pretrained GloVe word embeddings to measure semantic similarity between words, discover related terms, and perform vector-based word analogy tasks using Python and Gensim.

```python
# Install gensim (Google Colab only)
!pip install -q gensim

import gensim.downloader as api

print("Downloading pretrained GloVe embeddings...")
model = api.load("glove-wiki-gigaword-100")
print("Model loaded successfully!")

# ----------------------------------------------------
# Example 1: Word Similarity
# ----------------------------------------------------
word1 = "cat"
word2 = "dog"

similarity = model.similarity(word1, word2)

print(f"\nSimilarity between '{word1}' and '{word2}': {similarity:.4f}")

# ----------------------------------------------------
# Example 2: Most Similar Words
# ----------------------------------------------------
print("\nWords most similar to 'computer':")

for word, score in model.most_similar("computer", topn=5):
    print(f"{word:15s} {score:.4f}")

# ----------------------------------------------------
# Example 3: Word Analogy
# ----------------------------------------------------
print("\nAnalogy:")

result = model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=5
)

print("king - man + woman ≈")

for word, score in result:
    print(f"{word:15s} {score:.4f}")
```
