Learn how to adapt a pretrained ResNet-50 model for a new image classification task by replacing its final classification layer. This notebook demonstrates a fundamental transfer learning technique used to customize pretrained computer vision models for downstream applications.

```python

import torch.nn as nn
from torchvision import models

# Load a pretrained ResNet-50 model
weights = models.ResNet50_Weights.DEFAULT
model = models.resnet50(weights=weights)

# The original final layer predicts 1000 ImageNet classes.
print(f"Original final layer: {model.fc}")

# Number of input features to the final layer
num_ftrs = model.fc.in_features
print(f"Number of input features: {num_ftrs}")

# Replace the classifier for a binary classification task
model.fc = nn.Linear(num_ftrs, 2)

print(f"Modified final layer: {model.fc}")
```
