Goal: Build a recurrent neural network (RNN) in PyTorch that learns word embeddings while performing a sequence-to-one text classification task, such as sentiment analysis. This conceptual notebook demonstrates how an embedding layer converts integer-encoded words into trainable dense vectors, how the RNN processes the embedded word sequence to capture contextual information, and how the final hidden state is mapped through a fully connected layer to produce the final prediction.

```python
import torch.nn as nn

class MyRNNWithLearnedEmbeddings(nn.Module):
 """
 A conceptual PyTorch RNN model that uses an embedding layer to process
 integer-encoded words before feeding them into the recurrent layer.
 """
 def __init__(self, vocab_size, embedding_dim, hidden_size, output_size):
  super().__init__()

  # self.embedding: The embedding layer acts as a trainable lookup table.
  # It takes an integer ID and returns a dense vector of size embedding_dim.
  # The vectors are learned automatically during training.
  self.embedding = nn.Embedding(vocab_size, embedding_dim)

  # self.rnn: The RNN layer processes the sequence of embedded vectors.
  # Input shape: (batch_size, sequence_length, embedding_dim)
  self.rnn = nn.RNN(embedding_dim, hidden_size, batch_first=True)

  # self.fc: A fully connected layer that produces the final prediction.
  self.fc = nn.Linear(hidden_size, output_size)

 def forward(self, x_word_ids):
  # x_word_ids: Tensor of integer word IDs
  # Shape: (batch_size, sequence_length)

  # Look up the learned embedding vector corresponding to each word ID.
  # Output shape: (batch_size, sequence_length, embedding_dim)
  embedded_x = self.embedding(x_word_ids)

  # **The RNN processes the sequence one time step at a time.**
  rnn_output, _ = self.rnn(embedded_x)

  # For a sequence-to-one task, use only the output from the final time step.
  final_output = self.fc(rnn_output[:, -1, :])

  return final_output
```
