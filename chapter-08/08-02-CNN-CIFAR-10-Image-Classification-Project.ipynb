Hands-On Project: Building, Training, and Evaluating a CNN on CIFAR-10. Build a complete convolutional neural network in PyTorch to classify images from the CIFAR-10 dataset. This project demonstrates image preprocessing and normalization, CNN architecture design, training with Cross-Entropy loss and the Adam optimizer, test-set evaluation, prediction probabilities, and visualization of filters learned by the first convolutional layer.

```python
# ============================================================
# Import PyTorch and computer vision libraries
# ============================================================

import torch
import torch.nn as nn              # Neural network layers and loss functions
import torch.optim as optim        # Optimization algorithms

import torchvision                 # Computer vision datasets
import torchvision.transforms as T # Image preprocessing utilities

# ============================================================
# Convert images to tensors and normalize RGB channels
# ============================================================

transform = T.Compose([
    T.ToTensor(),                  # Convert image to a PyTorch tensor
    T.Normalize(
        (0.5, 0.5, 0.5),           # Mean for R, G, B channels
        (0.5, 0.5, 0.5)            # Standard deviation for R, G, B channels
    )
])

# ============================================================
# Dataset and DataLoader placeholders
# ============================================================

# train_loader = ...
# test_loader  = ...

# ============================================================
# Simple CNN model
# ============================================================

class SimpleCNN(nn.Module):

    def __init__(self):
        super().__init__()

        # Learn local visual patterns
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)

        # Reduce spatial dimensions
        self.pool = nn.MaxPool2d(2, 2)

        # Introduce non-linearity
        self.relu = nn.ReLU()

        # Convert learned features into class scores
        self.fc1 = nn.Linear(128 * 4 * 4, 256)
        self.fc2 = nn.Linear(256, 10)

    def forward(self, x):

        # Convolution Block 1
        x = self.pool(self.relu(self.conv1(x)))

        # Convolution Block 2
        x = self.pool(self.relu(self.conv2(x)))

        # Convolution Block 3
        x = self.pool(self.relu(self.conv3(x)))

        # Flatten feature maps into one vector
        x = x.view(x.size(0), -1)

        # Fully connected classifier
        x = self.relu(self.fc1(x))
        x = self.fc2(x)

        return x


model = SimpleCNN()

# ============================================================
# Loss function and optimizer
# ============================================================

# CrossEntropyLoss expects raw logits
criterion = nn.CrossEntropyLoss()

# Adam optimizer updates model weights during training
optimizer = optim.Adam(model.parameters(), lr=1e-3)

# ============================================================
# Training loop (placeholder)
# ============================================================

# for epoch in range(NUM_EPOCHS):
#
#     model.train()
#
#     for images, labels in train_loader:
#
#         optimizer.zero_grad()
#
#         logits = model(images)
#
#         loss = criterion(logits, labels)
#
#         loss.backward()
#
#         optimizer.step()
#
#     print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")

# ============================================================
# Evaluation (placeholder)
# ============================================================

# model.eval()
#
# correct = 0
# total = 0
#
# with torch.no_grad():
#
#     for images, labels in test_loader:
#
#         logits = model(images)
#
#         predictions = logits.argmax(dim=1)
#
#         correct += (predictions == labels).sum().item()
#
#         total += labels.size(0)
#
# accuracy = 100 * correct / total
#
# print(f"Test Accuracy: {accuracy:.2f}%")

# ============================================================
# Visualizing learned filters (placeholder)
# ============================================================

# def visualize_filters(conv_layer):
#     pass
#
# visualize_filters(model.conv1)

# ============================================================
# Converting logits into probabilities (teaching example)
# ============================================================

# During training, do NOT apply Softmax manually.
# CrossEntropyLoss expects raw logits.
# Softmax is used only for interpreting predictions.

# logits = model(images[:1])
#
# probabilities = torch.softmax(logits, dim=1)
#
# prediction = probabilities.argmax(dim=1).item()
#
# print("Predicted class:", prediction)
# print("Probability:", probabilities[0, prediction].item())
```
