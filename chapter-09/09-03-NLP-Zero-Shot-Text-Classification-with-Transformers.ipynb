Learn how to perform zero-shot text classification using Hugging Face Transformers without training a custom model. This notebook demonstrates how pretrained language models can categorize text into user-defined labels by leveraging natural language inference.

```python
# Step 1: Import the necessary library
# The `pipeline` function from the `transformers` library is a high-level API
# that simplifies using pre-trained models for various tasks.
from transformers import pipeline

# Step 2: Define the Problem and the Goal
print("Goal: Categorize a customer message into predefined labels using zero-shot classification.")
print("This shows how a model can perform a task without being explicitly trained on it.")

# Step 3: Load the zero-shot classification pipeline
# This line creates a `zero-shot-classification` pipeline.
# It automatically downloads a pre-trained model (in this case, facebook/bart-large-mnli)
# that is well-suited for this task.
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
print("Zero-shot classification pipeline loaded successfully.")

# Step 4: Define the text to be classified and the possible labels
# The `sequence` is the customer's message we want to categorize.
# The `candidate_labels` is a list of potential categories we want the model to choose from.
sequence_to_classify = "The app crashes every time I open it."
candidate_labels = ["praise", "complaint", "bug report", "feature request"]
print(f"\nText to classify: '{sequence_to_classify}'")
print(f"Candidate labels: {candidate_labels}")

# Step 5: Run the classifier and print the result
# The `classifier` object is called with the text and labels.
# It returns a dictionary containing the sorted labels and their corresponding confidence scores.
print("\nRunning the classifier...")
result = classifier(sequence_to_classify, candidate_labels)

# The result is a dictionary. We can access its keys to get the information we need.
print("Classification Result:")
# The 'sequence' key returns the original input text.
print(f"  - Original sequence: {result['sequence']}")
# The 'labels' key returns the candidate labels, sorted by confidence score (highest first).
print(f"  - Predicted labels: {result['labels']}")
# The 'scores' key returns the confidence score for each label in the same sorted order.
print(f"  - Confidence scores: {[f'{s:.2f}' for s in result['scores']]}")
print("\n")

# Step 6: Another example
# Let's try another message to show how the model adapts to a different context.
sequence_to_classify_2 = "It would be great if I could schedule deliveries."
print(f"Text to classify: '{sequence_to_classify_2}'")
print(f"Candidate labels: {candidate_labels}")
result_2 = classifier(sequence_to_classify_2, candidate_labels)
print("\nRunning the classifier...")
print("Classification Result:")
print(f"  - Original sequence: {result_2['sequence']}")
print(f"  - Predicted labels: {result_2['labels']}")
print(f"  - Confidence scores: {[f'{s:.2f}' for s in result_2['scores']]}")
```
